Importing the dependencies



```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score  
from sklearn.preprocessing import StandardScaler
```

Importing the datasets

```python
train_data = pd.read_csv('Datasets/train.csv')
test_data = pd.read_csv('Datasets/test.csv')
```

```python
train_data.head()
```

```python
train_data.isnull().sum()
```

Encoding the data 

```python
encoder = LabelEncoder()

for i in train_data.columns:
    if train_data[i].dtypes == "object":    
        train_data[i] = encoder.fit_transform(train_data[i].astype(str))
        test_data[i] = encoder.fit_transform(test_data[i].astype(str))
        
```

```python
train_data.head()
```

```python
train_data['loan_paid_back'] = train_data['loan_paid_back'].astype(int)
```

```python
train_data['loan_paid_back']
```

```python
train_data.drop(columns=['id'],inplace=True)
test_data.drop(columns=['id'],inplace=True)  
```

```python
scaler = StandardScaler()
```

```python
for i in train_data.columns:
    if i != 'loan_paid_back':
        train_data[i] = scaler.fit_transform(train_data[[i]])    

for i in test_data.columns:
    test_data[i] = scaler.fit_transform(test_data[[i]])
```

```python
test_data
```

Splitting the data

```python
X = train_data.drop(columns=['loan_paid_back'])
Y = train_data['loan_paid_back']
```

Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

Model Implementation

```python
model = LogisticRegression()

model.fit(X_train, Y_train)
```

```python
model.predict(X_test)
```

Getting the prediction for the test dataset

```python
test_predictions = model.predict(test_data)
```

Submitting the sample submission

```python
sample = pd.read_csv('Datasets/sample_submission.csv')
```

```python
Final_prediction = pd.DataFrame({
    'id': sample['id'][:len(test_predictions)],  
    'Price': test_predictions
})
```

Converting into the submission file

```python
Final_prediction.to_csv("Datasets/Final_Submission.csv", index=False)
```
